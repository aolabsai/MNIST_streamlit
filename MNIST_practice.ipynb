```python
#loading packages -->

from arch__MNIST import arch_bw, arch_gr
import ao_core as ao
import data_prep as data
import numpy as np
```

```python
# lets start with black and white and mnist 
agent = ao.Agent(arch_bw, notes="B&W MNIST Agent", save_meta=False)
```

```python
# the database we have. We can try pytorch mnist dataset as well. 
train_X = data.MN_TRAIN
train_Y = data.MN_TRAIN_Z
test_X = data.MN_TEST
test_Y = data.MN_TEST_Z
```

```python
n_train = 100 # define number  of images for training 
n_test = 10 # define number of images for testing


# randomly selecting some images  
np.random.seed(42)
indices_train = np.random.choice(train_X.shape[0], size=n_train, replace=False)
indices_test = np.random.choice(test_X.shape[0], size=n_test, replace=False)

train_X_sample = train_X[indices_train]
train_Y_sample = train_Y[indices_train]
test_X_sample = test_X[indices_test]
test_Y_sample = test_Y[indices_test]


TRAIN_INPUT = data.down_sample(train_X_sample).reshape(n_train, 784)  
TEST_INPUT = data.down_sample(test_X_sample).reshape(n_test, 784)
```

```python
TEST_INPUT.shape
```

```python
# tarining --> 

for i in range(len(TRAIN_INPUT)):
    agent.reset_state()
    agent.next_state(INPUT=TRAIN_INPUT[i], LABEL=train_Y_sample[i],  unsequenced=True)   
```

```python
# testing -->
correct = 0
for i in range(len(TEST_INPUT)):
    op = agent.next_state(INPUT=TEST_INPUT[i])
    op = op.detach().cpu().numpy()
    if np.array_equal(op.astype(float), test_Y_sample[i]):
        correct += 1
    # if i%5 == 0:    
    #     print('Testing going on ...' + str(i))    
print('Correctly predicted: ' + str(correct) + ' out of: ' + str(len(TEST_INPUT)))         
```
